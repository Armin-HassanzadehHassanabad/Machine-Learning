```python
import numpy as np
import pandas as pd
import io
from google.colab import files
```

```python
uploaded = files.upload()
df = pd.read_csv(io.BytesIO(uploaded[list(uploaded.keys())[0]]))
```

```python
df.head()
```

```python
X = df.drop("c", axis=1)
y = df["c"]
```

```python
X.fillna(0, inplace=True)
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import cross_val_score

MLP = MLPClassifier()
scores = cross_val_score(MLP, X, y, cv=10, scoring='accuracy')
scores.mean()
```

```python
MLP = MLPClassifier(activation='tanh')
scores = cross_val_score(MLP, X, y, cv=10, scoring='accuracy')
scores.mean()
```

```python
MLP = MLPClassifier(activation='relu', random_state=8)
scores = cross_val_score(MLP, X, y, cv=10, scoring='accuracy')
scores.mean()
```

```python
MLP = MLPClassifier(activation='relu', hidden_layer_sizes=(80, 100))
scores = cross_val_score(MLP, X, y, cv=10, scoring='accuracy')
scores.mean()
```

```python
MLP = MLPClassifier(activation='relu',
                    hidden_layer_sizes=(80, 100),
                    learning_rate_init=0.05)
scores = cross_val_score(MLP, X, y, cv=10, scoring='accuracy')
scores.mean()
```

```python
MLP = MLPClassifier(activation='relu',
                    hidden_layer_sizes=(80, 100),
                    learning_rate_init=0.05,
                    learning_rate='invscaling')
scores = cross_val_score(MLP, X, y, cv=10, scoring='accuracy')
scores.mean()
```

```python
MLP = MLPClassifier(activation='relu',
                    hidden_layer_sizes=(80, 100),
                    learning_rate_init=0.05,
                    learning_rate='invscaling',
                    momentum=0.9)
scores = cross_val_score(MLP, X, y, cv=10, scoring='accuracy')
scores.mean()
```

```python
MLP = MLPClassifier(activation='relu',
                    hidden_layer_sizes=(80, 100),
                    learning_rate_init=0.05,
                    learning_rate='invscaling',
                    momentum=0.9,
                    max_iter=50)
scores = cross_val_score(MLP, X, y, cv=10, scoring='accuracy')
scores.mean()
```

```python
MLP = MLPClassifier(activation='relu',
                    hidden_layer_sizes=(80, 100),
                    learning_rate_init=0.05,
                    learning_rate='invscaling',
                    momentum=0.9,
                    max_iter=50,
                    verbose=1)
scores = cross_val_score(MLP, X, y, cv=10, scoring='accuracy')
scores.mean()
```

```python
from sklearn.model_selection import train_test_split

Xtrain, Xtest, Ytrain, Ytest = train_test_split(X, y, test_size=0.25, random_state=7)

MLP = MLPClassifier(activation='relu', hidden_layer_sizes=(80, 100),
                    learning_rate_init=0.05,
                    learning_rate='invscaling',
                    momentum=0.9, max_iter=50)

MLP.fit(Xtrain, Ytrain)
pred = MLP.predict(Xtest)
pred
```

```python
pred = MLP.predict_proba(Xtest)
pred
```

```python
from sklearn.model_selection import GridSearchCV

MLP = MLPClassifier(random_state=42)

param = {"activation": ["relu", "logistic", "tanh"],
         "hidden_layer_sizes": [(10), (20), (20,30)],
         "max_iter": [100, 200, 300, 400],
         "learning_rate_init": [0.01, 0.001, 0.025]}

GS = GridSearchCV(MLP, param, cv=10).fit(X, y)
```
