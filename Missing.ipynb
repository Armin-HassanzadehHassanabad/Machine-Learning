```python
import pandas as pd
import numpy as np
from google.colab import files
import io
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
uploaded = files.upload()
df = pd.read_csv(io.BytesIO(uploaded[list(uploaded.keys())[0]]))
```

```python
df.isnull().sum()
```

```python
df.isnull().mean() * 100
```

```python
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values Heatmap")
plt.show()
```

```python
df_ex = df.fillna(df.mean())
df_ex.isnull().sum()
```

```python
df_explain = df.copy()
df_explain[['cholestoral ', 'blood sugar']] = df_explain[['cholestoral ', 'blood sugar']].fillna(df_explain[['cholestoral ', 'blood sugar']].mean())
df_explain.isnull().sum()
```

```python
df_ex = df.dropna(axis=1)
print(df.shape, df_ex.shape)
```

```python
df_ex = df.dropna(thresh=400, axis=1)
print(df.shape, df_ex.shape)
```

```python
df_ex = df.dropna(thresh=11, axis=0)
print(df.shape, df_ex.shape)
```

```python
df_explain['thal'] = df_explain['thal'].fillna(df_explain['thal'].mode()[0])
print(df['thal'].isnull().sum())
print(df['thal'].value_counts(dropna=False))
```

```python
df.ffill()
df.bfill()
```

```python
df_missing_corr = df.isnull().corr()
sns.heatmap(df_missing_corr, cmap="coolwarm")
plt.title("Missingness Correlation Matrix")
```
