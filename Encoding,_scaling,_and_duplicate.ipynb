```python
import pandas as pd
import numpy as np
import io
from google.colab import files
```

```python
uploaded = files.upload()
df = pd.read_csv(io.BytesIO(uploaded[list(uploaded.keys())[0]]))
```

**Label Encoding**

```python
df['sex'][df['sex'] == 1] = 'M'
df['sex'][df['sex'] == 0] = 'F'
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder


LE = LabelEncoder()
LE.fit(df['sex'])
df['sex'] = LE.transform(df['sex'])
df.head()
```

```python
df.loc[3, 'sex'] = np.nan
df_dummy = pd.get_dummies(df, dummy_na=True, drop_first=True)
df_dummy.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

rsc = MinMaxScaler()
df_scaled = pd.DataFrame(rsc.fit_transform(df), columns=df.columns, index=df.index)
df_scaled.head()
```
